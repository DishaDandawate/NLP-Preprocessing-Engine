## Task 1: Conceptual Understanding 

### Q1. What is the difference between "Love" and "love" in NLP?
#### ANS: In NLP, "Love" and "love" are treated as different words because of case sensitivity. This can create duplicate words in data. So, we usually convert all text into lowercase to make it consistent.

### Q2. What happens if stopwords are not removed?
#### ANS: If stopwords are not removed, the data will contain many unnecessary words like "is", "the", "am". This increases noise and can reduce the performance and accuracy of the model.

### Q3. Provide two real-world scenarios where removing stopwords can be harmful.
#### ANS:- In sentiment analysis, removing words like "not" can change the meaning. For example, "not good" becomes "good".
#### - In question-answering systems or chatbots, removing words like "what", "how", "when" can make the sentence meaningless.

### Q4. Explain the difference between stemming and lemmatization.
#### ANS: Stemming simply cuts the word to its root form, which may not always be meaningful. For example, "playing" becomes "play".
#### Lemmatization converts a word into its correct base form using meaning. For example, "better" becomes "good". It is more accurate than stemming.

## Task 2:Build Advanced Preprocessing Function


```python
import re
def preprocess_text(text):
    if not text:
        return [], ""
    
    text = text.lower()
    text = re.sub(r'http\S+|www\S+|@\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    
    tokens = text.split()
    tokens = [word for word in tokens if len(word) > 2 or word in ['no', 'not']]
    
    cleaned_sentence = " ".join(tokens)
    
    return tokens, cleaned_sentence
```

## Task 3: Stress Testing

```python
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

for s in sentences:
    tokens, clean = preprocess_text(s)
    print("Original:", s)
    print("Tokens:", tokens)
    print("Cleaned:", clean)
    print("-"*40)
```

## Task 4: Token Analytics

```python
def analyze(tokens):
    if len(tokens) == 0:
        return 0, 0, 0
    
    total = len(tokens)
    unique = len(set(tokens))
    avg_len = sum(len(w) for w in tokens) / total
    
    return total, unique, avg_len

for s in sentences:
    tokens, _ = preprocess_text(s)
    t, u, a = analyze(tokens)
    print(s)
    print(f"Total: {t}, Unique: {u}, Avg Length: {a:.2f}")
    print("-"*40)
```

#### 1. Which sentence had the most noise?
##### Ans: "Win $$$ now!!! Limited offer!!!" has the most noise because it contains symbols, special characters, and unnecessary punctuation.
#### 2. Which sentence retained the most meaningful tokens after cleaning?
##### Ans: "I am not happy with this" retained the most meaningful tokens.

## Task 5: Frequency Analysis


```python
from collections import Counter
all_tokens = []
for s in sentences:
    tokens, _ = preprocess_text(s)
    all_tokens.extend(tokens)

counter = Counter(all_tokens)

print("Top 10 frequent words:", counter.most_common(10))
print("Least 5 frequent words:", counter.most_common()[-5:])
```

## Task 6: Build Full Pipeline


```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []
    
    for text in text_list:
        tokens, clean = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean)
    
    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

result = full_pipeline(sentences)
print(result)
```
